## Data Preprocessing

````python
import pandas as pd
file_path = "/content/drive/MyDrive/combined_data_1.txt"

# Initialize an empty list to store processed rows.
rows=[]
# Open the raw data file for reading.
with open(file_path,'r') as f:
  for line in f:
    line = line.strip() # Remove leading/trailing whitespace from each line.

    # Check if the line ends with a colon, indicating a Movie ID entry.
    if line.endswith(':'):
      rows.append([line[:],None]) # Store the Movie ID and None for the Rating (as it's not a rating line).
    else:
      # If it's not a Movie ID, it's a UserID, Rating, and Date entry.
      parts = line.split(',') # Split the line by comma.
      rows.append(parts[:2]) # Store only UserID and Rating, ignoring the Date for this project.
# Create a Pandas DataFrame from the processed rows, with specified column names.
df = pd.DataFrame(rows,columns=['MovieID/UserID','Rating'])
# Display the first few rows of the newly created DataFrame to verify correct parsing.
df.head()
````

````python
# Initialize an empty list to hold the Movie_id for each row.
movies_list = []
# Initialize movie_id to None; it will store the current movie ID being processed.
movie_id = None

# Iterate through the 'MovieID/UserID' column of the DataFrame.
for id in df["MovieID/UserID"]:
  # Check if the current entry 'id' contains a colon, which indicates it's a Movie ID marker.
  if ":" in id: # If 'id' is a Movie ID (e.g., "1:")
    id= id.replace(":","") # Remove the colon to get the clean Movie ID.
    movie_id = id # Update the current 'movie_id'.
  # Append the current 'movie_id' to the 'movies_list'. This way, each rating gets its corresponding movie ID.
  movies_list.append(movie_id)

# Display the generated list of movie IDs. (This output can be very long for large datasets).
movies_list
````

````python
# Add the 'movies_list' as a new column named 'Movie_id' to the DataFrame.
# This associates each user's rating with the correct movie.
df["Movie_id"] = movies_list
````

````python
# Drop rows where the 'Rating' column is NaN (None). These are the original Movie ID marker rows.
# 'inplace=True' modifies the DataFrame directly without needing to reassign it.
df.dropna(inplace=True)
````

````python
# Rename the 'MovieID/UserID' column to 'User_id' for better clarity, as it now exclusively contains user IDs.
df.rename(columns={"MovieID/UserID":"User_id"},inplace=True)
# Display the first few rows of the DataFrame to show the renamed column.
df.head()
````

````python
# Calculate the total number of unique movies in the dataset.
total_movies = df["Movie_id"].nunique()
# Display the total count of unique movies.
total_movies
````

````python
# Count the occurrences of each unique 'Movie_id' to determine how many ratings each movie received.
movie_counts = df["Movie_id"].value_counts()
# Display the series of movie counts, showing movies and their respective total ratings.
movie_counts # Number of ratings given to each movie
````

````python
# Calculate the 60th percentile of the movie rating counts. Movies with fewer ratings than this will be considered 'less rated'.
threshold_movie = np.percentile(movie_counts, 60)
# Display the calculated threshold value.
threshold_movie
````

````python
# Calculate the total number of unique users in the dataset.
total_users= df["User_id"].nunique()
# Display the total count of unique users.
total_users
````

````python
# Count the occurrences of each unique 'User_id' to determine how many movies each user rated.
user_counts = df["User_id"].value_counts()
# Display the series of user counts, showing users and their respective total ratings.
user_counts # Number of movies rated by each user
````

````python
# Calculate the 60th percentile of the user rating counts. Users who rated fewer movies than this will be considered 'inactive'.
threshold_user = np.percentile(user_counts, 60)
# Display the calculated threshold value.
threshold_user
````

````python
# Identify movie IDs where the rating count is below the 'threshold_movie'.
# '.index' extracts the movie IDs (as they are the index of the 'movie_counts' series).
drop_movies = movie_counts[movie_counts<threshold_movie].index
# Display the index of movies to be dropped.
drop_movies
````

````python
# Identify user IDs where the number of rated movies is below the 'threshold_user'.
# '.index' extracts the user IDs (as they are the index of the 'user_counts' series).
drop_users = user_counts[user_counts<threshold_user].index
# Display the index of users to be dropped.
drop_users
````

````python
# This line checks if the string "1733" is present in the 'drop_movies' index.
# The 'not' operator inverts the boolean result.
not("1733" in drop_movies)
````

````python
# Filter the DataFrame 'df' to remove rows corresponding to movies in 'drop_movies'.
# '~' negates the boolean series, keeping only the movies *not* in 'drop_movies'.
df = df[~df["Movie_id"].isin(drop_movies)]
````

````python
# Filter the DataFrame 'df' to remove rows corresponding to users in 'drop_users'.
# '~' negates the boolean series, keeping only the users *not* in 'drop_users'.
df = df[~df["User_id"].isin(drop_users)]
````

````python
# Convert the 'User_id' column to integer type.
df["User_id"]=df["User_id"].astype(int)
# Convert the 'Rating' column to float type, as ratings can be fractional.
df["Rating"]=df["Rating"].astype(float)
# Convert the 'Movie_id' column to integer type.
df["Movie_id"]= df["Movie_id"].astype(int)
# Display the data types of each column to confirm the conversions.
df.dtypes
````

## Model Training

````python
# Install the 'scikit-surprise' library, which provides algorithms and tools for recommender systems.
!pip install scikit-surprise
````

````python
# Import the 'Reader' and 'Dataset' classes from the 'surprise' library.
# 'Reader' helps parse data, and 'Dataset' is used to load data into a format compatible with 'surprise'.
from surprise import Reader, Dataset
````

````python
# Import the Singular Value Decomposition (SVD) algorithm from the 'surprise' library.
# SVD is a matrix factorization technique widely used in collaborative filtering.
from surprise import SVD
````

````python
# Initialize a 'Reader' object. It's configured to understand that ratings are on a scale of 1 to 5.
reader = Reader(rating_scale=(1,5))
````

````python
# Load the preprocessed DataFrame into a 'surprise' Dataset object.
# We select only the 'User_id', 'Movie_id', and 'Rating' columns, and limit to the first 100,000 rows for demonstration/speed.
data = Dataset.load_from_df(df[["User_id","Movie_id","Rating"]][:100000],reader)
# Display the Dataset object.
data
````

````python
# Build a full training set from the loaded 'surprise' Dataset.
# This object will be used to train the recommendation model.
trainset = data.build_full_trainset()
# Display the Trainset object.
trainset
````

````python
# Initialize the SVD (Singular Value Decomposition) model.
model = SVD()
# Train the SVD model on the 'trainset'. This process involves matrix factorization to learn user and item latent features.
model.fit(trainset)
# Display the trained SVD model object.
model
````

````python
# Make a prediction for a specific user (uid=1331154) and a specific movie (iid=1).
# This demonstrates how the trained model can estimate a rating for a user-item pair.
pred = model.predict(uid = 1331154, iid =1)

# Print the estimated rating from the prediction object.
print("Estimated Rating =",pred.est)
````

## Recommendation Generation

### Get movies the user has not watched

````python
user_id_to_recommend = 1331154

# Get a list of all unique movie IDs present in the filtered dataset.
all_movie_ids = df['Movie_id'].unique()

# Get a list of movie IDs that the target user has already watched.
# Filter the DataFrame for the specific user and then get unique 'Movie_id' values.
watched_movies = df[df['User_id'] == user_id_to_recommend]['Movie_id'].unique()

# Identify movies not watched by the user by finding the difference between all movies and watched movies.
unwatched_movies = [movie_id for movie_id in all_movie_ids if movie_id not in watched_movies]

# Print statistics about the movie sets for verification.
print(f"Total movies: {len(all_movie_ids)}")
print(f"Movies watched by user {user_id_to_recommend}: {len(watched_movies)}")
print(f"Movies not watched by user {user_id_to_recommend}: {len(unwatched_movies)}")
````

### Predict ratings for unwatched movies

````python
predictions = []
# Iterate through each movie that the target user has not yet watched.
for movie_id in unwatched_movies:
    # Predict the rating the user would give to the current unwatched movie.
    pred = model.predict(uid=user_id_to_recommend, iid=movie_id)
    # Append a tuple of (movie_id, estimated_rating) to the predictions list.
    predictions.append((movie_id, pred.est))

# Sort the list of predictions by the estimated rating in descending order.
# The lambda function specifies to sort based on the second element of each tuple (the rating).
predictions.sort(key=lambda x: x[1], reverse=True)

# Extract the Movie IDs of the top 10 recommended movies.
top_10_movie_ids = [movie_id for movie_id, _ in predictions[:10]]

# Print a header for the recommended movies.
print("Top 10 recommended movie IDs:")
# Iterate through the top 10 predictions and print each movie ID along with its predicted rating, formatted to two decimal places.
for movie_id, rating in predictions[:10]:
    print(f"Movie ID: {movie_id}, Predicted Rating: {rating:.2f}")
````

### Display the top 10 recommended movies with titles

````python
# Create a Pandas DataFrame from the top 10 predictions, with columns 'Movie_id' and 'Predicted_Rating'.
recommended_movies_df = pd.DataFrame(predictions[:10], columns=['Movie_id', 'Predicted_Rating'])
# Merge this DataFrame with the 'movies' DataFrame (which contains movie titles).
# The merge is performed on 'Movie_id' from our predictions and 'movieId' from the movies metadata.
# 'how='left'' ensures all recommended movies are kept, and matching titles are added.
recommended_movies_with_titles = pd.merge(recommended_movies_df, movies, left_on='Movie_id', right_on='movieId', how='left')

# Display the final DataFrame, showing the Movie ID, Title, and Predicted Rating for the top 10 recommendations.
display(recommended_movies_with_titles[['Movie_id', 'title', 'Predicted_Rating']])
````
