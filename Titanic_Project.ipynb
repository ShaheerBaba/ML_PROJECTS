# Titanic Survival Prediction Project

This notebook demonstrates a complete machine learning workflow to predict passenger survival on the Titanic, including data loading, preprocessing, model training, and evaluation using various classification algorithms.

## 1. Data Loading and Initial Exploration

In this section, we load the Titanic dataset and perform an initial exploration to understand its structure, identify missing values, and check for duplicates.

```python
df = sns.load_dataset("titanic")
```

## 2. Data Preprocessing

This section focuses on cleaning and transforming the data to prepare it for machine learning models. This involves handling missing values, dropping irrelevant columns, and encoding categorical features.

## 3. Data Splitting and Scaling

Here, we split the dataset into training and testing sets, and then scale the numerical features to ensure that all features contribute equally to the model training process.

```python
x = df.drop('survived',axis=1)
```

```python
y = df["survived"]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.fit_transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.linear_model import LogisticRegression
# Initialize the Logistic Regression model
# random_state ensures reproducibility of results
# max_iter is increased to ensure the optimization algorithm converges
model = LogisticRegression(random_state=42, max_iter=200)
```

```python
# Train the Logistic Regression model using the scaled training data
# The model learns the relationship between features (x_train_scaled) and the target (y_train)
model.fit(x_train_scaled,y_train)
```

```python
# Make predictions on the scaled test data using the trained Logistic Regression model
y_pred = model.predict(x_test_scaled)
```

```python
# Display the predictions made by the Logistic Regression model for the test set
# These are the predicted 'survived' (0 or 1) values
y_pred
```

### 4.1 Logistic Regression

First, we train a Logistic Regression model, a simple yet effective linear classifier.

```python
from sklearn.preprocessing import StandardScaler

# Initialize the scaler
scaler = StandardScaler()

# Fit the scaler on the training data and transform both training and testing data
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test) # Use transform for test set to prevent data leakage
```

## 4. Model Training and Evaluation

In this section, we train and evaluate several classification models on the preprocessed data. We will compare their performance using accuracy, confusion matrix, and classification reports.

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
from sklearn.neighbors import KNeighborsClassifier
# Initialize the K-Nearest Neighbors model with 5 neighbors
# 'n_neighbors' specifies the number of neighbors to consider for classification
knn_model = KNeighborsClassifier(n_neighbors=5)
```

```python
# Train the KNN model using the scaled training data
# The model 'learns' by storing the training data points
knn_model.fit(x_train_scaled,y_train)
```

```python
# Make predictions on the scaled test data using the trained KNN model
y_pred_knn = knn_model.predict(x_test_scaled)
```

```python
# Display the predictions made by the KNN model for the test set
# These are the predicted 'survived' (0 or 1) values from the KNN model
y_pred_knn
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
print("Accuracy score :",accuracy_score(y_test,y_pred_knn))
```

```python
print("confusion matrix :",confusion_matrix(y_test,y_pred_knn))
```

```python
print(classification_report(y_test,y_pred_knn))
```

### 4.2 K-Nearest Neighbors (KNN)

Next, we train a K-Nearest Neighbors classifier, a non-parametric method used for classification and regression.

```python
# Import Gaussian Naive Bayes model
from sklearn.naive_bayes import GaussianNB

# Initialize the Gaussian Naive Bayes model
model_NB = GaussianNB()
```

```python
# Train the Naive Bayes model using the original (unscaled) training data
# Gaussian Naive Bayes makes assumptions about feature distributions but typically doesn't require scaling.
model_NB.fit(x_train,y_train)
```

```python
# Make predictions on the original (unscaled) test data using the trained Naive Bayes model
y_pred_NB = model_NB.predict(x_test)
```

```python
# Display the predictions made by the Naive Bayes model for the test set
# These are the predicted 'survived' (0 or 1) values from the Naive Bayes model
y_pred_NB
```

```python
print(accuracy_score(y_test,y_pred_NB))
print(confusion_matrix(y_test,y_pred_NB))
print(classification_report(y_test,y_pred_NB))
```

### 4.3 Gaussian Naive Bayes

Then, we use a Gaussian Naive Bayes classifier, which assumes features are normally distributed.

```python
# Import DecisionTreeClassifier model
from sklearn.tree import DecisionTreeClassifier

# Initialize the Decision Tree model with a random state for reproducibility
# random_state ensures that the splits in the tree are consistent across runs
model_DT = DecisionTreeClassifier(random_state=42)
```

### 4.4 Decision Tree Classifier

We also train a Decision Tree, a non-linear model that partitions the data based on feature values.

```python
from sklearn.svm import SVC
```

```python
from sklearn.svm import SVC
# Initialize the Support Vector Machine (SVM) model with a radial basis function (rbf) kernel
# The 'rbf' kernel is good for non-linear decision boundaries.
# random_state ensures reproducibility.
model_SVM = SVC(kernel = "rbf", random_state=42)
```

```python
# Train the SVM model using the scaled training data
# SVMs are sensitive to feature scaling, so using scaled data is crucial for optimal performance.
model_SVM.fit(x_train_scaled,y_train)
```

```python
# Make predictions on the scaled test data using the trained SVM model
y_pred_SVM = model_SVM.predict(x_test_scaled)
```

```python
# Display the predictions made by the SVM model for the test set
# These are the predicted 'survived' (0 or 1) values from the SVM model
y_pred_SVM
```

```python
print(accuracy_score(y_test,y_pred_SVM))
print(confusion_matrix(y_test,y_pred_SVM))
print(classification_report(y_test,y_pred_SVM))
```
